```python
import pandas as pd
import numpy as np
```

```python


data = {
    "Customer_ID": [101, 102, 103, 104, 105, 106, 106, 108, 109, 110],
    "Name": ["Amit", "Riya", "Rahul", "Priya", "Amit", "Neha", "Neha", "Karan", "Simran", "Arjun"],
    "Age": [25, 30, np.nan, 28, 25, 35, 35, -5, 29, 200],
    "City": ["Delhi", "mumbai", "DELHI", "Pune", "Delhi", "Jaipur", "Jaipur", "pune", None, "Mumbai"],
    "Purchase_Amount": [1500, 2500, 1800, np.nan, 1500, 3200, 3200, 900, -500, 4100]
}

df = pd.DataFrame(data)

print("Original Customer Dataset")
display(df)

print("Dataset Shape:", df.shape)
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Data Quality Analysis

print("===== DATA QUALITY REPORT =====")

print("\n1. Dataset Information")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\n2. Missing Values")
print(df.isnull().sum())

print("\n3. Duplicate Rows")
print("Duplicate rows:", df.duplicated().sum())

print("\n4. Invalid Age Records")
print(df[(df["Age"] < 0) | (df["Age"] > 100)])

print("\n5. Negative Purchase Amounts")
print(df[df["Purchase_Amount"] < 0])

print("\n6. Data Types")
print(df.dtypes)
```

```python
# Create a copy of the original dataset
clean_df = df.copy()

# 1. Remove exact duplicate rows
clean_df = clean_df.drop_duplicates()

# 2. Standardize city names
clean_df["City"] = clean_df["City"].str.strip().str.title()

# 3. Replace invalid ages with missing values
clean_df.loc[
    (clean_df["Age"] < 0) | (clean_df["Age"] > 100), "Age"
] = np.nan

# 4. Replace negative purchase amounts with missing values
clean_df.loc[
    clean_df["Purchase_Amount"] < 0, "Purchase_Amount"
] = np.nan

# 5. Fill missing ages with the median
clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].median()
)

# 6. Fill missing purchase amounts with the median
clean_df["Purchase_Amount"] = clean_df["Purchase_Amount"].fillna(
    clean_df["Purchase_Amount"].median()
)

# Display cleaned data
print("Cleaned Dataset")
display(clean_df)

print("Original shape:", df.shape)
print("Cleaned shape:", clean_df.shape)
```

```python
# Save the cleaned dataset as a CSV file
clean_df.to_csv("cleaned_customer_data.csv", index=False)

print("File saved successfully!")
print("Filename: cleaned_customer_data.csv")
```

```python
# Professional Data Quality Report

report = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Total columns",
        "Exact duplicate rows",
        "Missing values (all columns)",
        "Invalid age records",
        "Negative purchase amounts",
        "Missing values after cleaning"
    ],
    "Before Cleaning": [
        len(df),
        len(df.columns),
        int(df.duplicated().sum()),
        int(df.isnull().sum().sum()),
        int(((df["Age"] < 0) | (df["Age"] > 100)).sum()),
        int((df["Purchase_Amount"] < 0).sum()),
        "Not applicable"
    ],
    "After Cleaning": [
        len(clean_df),
        len(clean_df.columns),
        int(clean_df.duplicated().sum()),
        int(clean_df.isnull().sum().sum()),
        int(((clean_df["Age"] < 0) | (clean_df["Age"] > 100)).sum()),
        int((clean_df["Purchase_Amount"] < 0).sum()),
        int(clean_df.isnull().sum().sum())
    ]
})

display(report)

# Save report as CSV
report.to_csv("data_quality_report.csv", index=False)

print("Data quality report saved successfully!")
```
